```python
### Step 1: Import Libraries
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder
from sklearn.neighbors import KNeighborsClassifier
from sklearn.linear_model import LogisticRegression
from sklearn.naive_bayes import GaussianNB
from sklearn.metrics import accuracy_score, classification_report

print("Libraries imported successfully")
```

```python
### Step 2: Load and Explore Dataset
df = pd.read_csv("Iris.csv")
print("Shape:", df.shape)
df.head()
```

```python
print(df['Species'].value_counts())
```

```python
### Step 3: Preprocess Data
df = df.drop('Id', axis=1)

X = df[['SepalLengthCm', 'SepalWidthCm', 'PetalLengthCm', 'PetalWidthCm']]
y = df['Species']

le = LabelEncoder()
y = le.fit_transform(y)

print("Classes:", le.classes_)
print("Encoded values: 0, 1, 2")
```

```python
### Step 4: Train-Test Split
X_train, _, y_train, _ = train_test_split(X, y, test_size=0.5, random_state=42)

X_test = X
y_test = y

print("Training samples:", X_train.shape[0])
print("Testing samples :", X_test.shape[0])
```

```python
### Step 5: Model 1 - K-Nearest Neighbours (KNN)
knn = KNeighborsClassifier(n_neighbors=5)
knn.fit(X_train, y_train)
knn_pred = knn.predict(X_test)

print("KNN Accuracy:", round(accuracy_score(y_test, knn_pred) * 100, 2), "%")
print()
print(classification_report(y_test, knn_pred, target_names=le.classes_))
```

```python
### Step 6: Model 2 - Logistic Regression
lr = LogisticRegression(max_iter=200)
lr.fit(X_train, y_train)
lr_pred = lr.predict(X_test)

print("Logistic Regression Accuracy:", round(accuracy_score(y_test, lr_pred) * 100, 2), "%")
print()
print(classification_report(y_test, lr_pred, target_names=le.classes_))
```

```python
### Step 7: Model 3 - Naive Bayes
nb = GaussianNB()
nb.fit(X_train, y_train)
nb_pred = nb.predict(X_test)

print("Naive Bayes Accuracy:", round(accuracy_score(y_test, nb_pred) * 100, 2), "%")
print()
print(classification_report(y_test, nb_pred, target_names=le.classes_))
```

```python
### Step 8: Model Comparison Summary
knn_acc = round(accuracy_score(y_test, knn_pred) * 100, 2)
lr_acc  = round(accuracy_score(y_test, lr_pred) * 100, 2)
nb_acc  = round(accuracy_score(y_test, nb_pred) * 100, 2)

results = pd.DataFrame({
    'Model': ['KNN', 'Logistic Regression', 'Naive Bayes'],
    'Accuracy (%)': [knn_acc, lr_acc, nb_acc]
})

results = results.sort_values('Accuracy (%)', ascending=False).reset_index(drop=True)
print(results)
print()
print("Best Model:", results.iloc[0]['Model'], "with", results.iloc[0]['Accuracy (%)'], "% accuracy")
```
